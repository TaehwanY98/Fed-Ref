```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display
from matplotlib.patches import Polygon

def draw_covariance_shadow(ax, x, y, y_std, color="b", alpha= 0.2,**kwargs):
    """공분산 행렬을 기반으로 N-표준편차 크기의 그림자 타원을 그리는 함수"""
    # 공분산 타원 객체 생성
    xy1 = [np.array([x1, y1]) for x1 , y1 in zip(x, y+y_std)]
    xy2 = [np.array([x1, y1]) for x1 , y1 in zip(x, y-y_std)]

    polygon = Polygon(
        xy= xy1+xy2,
        closed=False, 
        facecolor=color,  # Interior color
        edgecolor=color, # Outline color
        linewidth=1           # Border thickness
    )
    ax.add_patch(polygon)
    return polygon


def newploting(fedavg, fedprox, fedref, title="MNIST Classification", metric=["loss"], num=60, thresholds=[0.5,0.7,1], legend= False, start=1):
    # 예시 데이터
    fedref_colors = ["blue", "orange", "green", "yellow", "khaki", "cyan", "lightcoral", "darkred", "tomato", "red"]
    labels = ["FedAvg", "FedProx (μ:0.5)", "FedOpt (η:0.01, β1:0.9, β2:0.99, τ:1e-4)", "AdaBest (α:0.1 , β:1.0)", "FedEve (β:0.9, μ:0.01, ρ:0.9)",
              "A-FedPD (μ:0.1)", "FedRef (λ:0.01, ρ:4) client: FedAvg" , "FedRef (λ:0.01, ρ:4) client: FedProx", "FedRef (λ:0.1, ρ:8) client: FedAvg"]
    
    with plt.style.context('bmh'):
        fig, ax = plt.subplots(1, len(metric), figsize=(24, 4))
        
        # metric이 1개일 때 ax가 배열이 아닌 단일 객체가 되는 것을 방지
        if len(metric) == 1:
            ax = [ax]
            
        for i, mtr in enumerate(metric):
            polygons = [] # 각 서브플롯마다 타원을 따로 수집
           
            # 1. FedAvg 선 및 타원 생성
            ax[i].plot(fedavg["dataframe"][mtr][start:num], marker=".", linestyle="--", color=fedref_colors[0], alpha=0.7, linewidth=4)
            # polygons.append(draw_covariance_shadow(ax[i], range(start, num), fedavg["dataframe"][mtr][start:num].values, fedavg["variances"][mtr][start:num].values,color=fedref_colors[0], n_std=1.0))
            
            # 2. FedProx 선 및 타원 생성
            ax[i].plot(fedprox["dataframe"][mtr][start:num], marker=".", linestyle="--", color=fedref_colors[1], alpha=0.7, linewidth=3)
            # polygons.append(draw_covariance_shadow(ax[i], range(start, num), fedprox["dataframe"][mtr][start:num], fedprox["variances"][mtr][start:num].values,color=fedref_colors[1], n_std=1.0))
            
            # 3. FedRef 루프 선 및 타원 생성
            for i2, l in enumerate(fedref):
                ax[i].plot(l["dataframe"][mtr][start:num], marker=".", color=fedref_colors[i2+2], linestyle="--", alpha=1, linewidth=3)
                # polygons.append(draw_covariance_shadow(ax[i], range(start, num), l["dataframe"][mtr][start:num], l["variances"][mtr][start:num].values,color=fedref_colors[i2+2], n_std=1.0))
                
            # 🔥 핵심 수정: 수집된 모든 타원을 현재 축(ax[i])에 추가
            
            # 메인 축 설정
            ax[i].set_xlabel("Rounds", fontsize=15)
            ax[i].autoscale(enable=True, axis='y')
            ax[i].set_ylabel(mtr, fontsize=15)
            ax[i].tick_params(axis='x', labelsize=15)
            ax[i].tick_params(axis='y', labelsize=15)

            
        h, l = ax[len(metric)//2].get_legend_handles_labels()
        for label in labels:
            l.append(label)
        h.extend([plt.Line2D([0], [0], color=fedref_colors[0], lw=4), plt.Line2D([0], [0], color=fedref_colors[1], lw=4)])
        for i in range(len(fedref)):
            h.append(plt.Line2D([0], [0], color=fedref_colors[i+2], lw=4))

            
        if legend == True:
            ax[len(metric)//2].legend(
                h, l,
                loc="upper center",       
                bbox_to_anchor=(0.5, -0.25), 
                ncols=2,                   
                fontsize=16
            )
    display(fig)

```

```python
import numpy as np

def newDataFrame(
    seeds=[],
    TYPE="fedavg",
    degraded=0.2,
    n=51,
    metrics=["loss", "accuracy", "f1score"],
):
    # 각 시드별 데이터를 담을 딕셔너리 구조: { "loss": [df_seed1, df_seed2, ...], ... }
    metric_data = {m: [] for m in metrics}

    # 1. 파일 읽기 및 metric별 Series 수집
    for seed in seeds:
        root_dir = f"Result/{TYPE}/degrade{degraded}/{seed}"
        file_path = f"{root_dir}/{TYPE}_cinic10_lda0.1_p4.csv"
        df = pd.read_csv(file_path)

        for m in metrics:
            # Series(1차원) 형태로 수집
            metric_data[m].append(df[m])

    # 2. 결과 저장용 데이터프레임 초기화
    mean_df = pd.DataFrame(index=range(n))
    std_df = pd.DataFrame(index=range(n))

    # 3. 각 metric별 평균 및 표준편차(분산 기반) 계산
    for m in metrics:
        # 각 seed의 Series를 열(Column)로 묶어 (51, len(seeds)) 형태의 DataFrame 생성
        combined = pd.concat(metric_data[m], axis=1)

        # 행(Row) 단위로 평균과 표준편차 계산 (1차원 1D Series 반환)
        mean_df[m] = combined.mean(axis=1)
        std_df[m] = combined.std(axis=1)  # 분산을 원하시면 .var(axis=1)로 변경

    result = {"dataframe": mean_df, "variances": std_df}

    return result
```

```python
n = 51
thresholds=[i for i in range(5, n, 5)]
seeds = [2024, 2025, 2026, 2027, 2028, 2029]

TYPE="fedavg"
fedavg = newDataFrame(seeds, TYPE, degraded=0.2)
# print(fedavg["dataframe"]["accuracy"]-fedavg["variances"]["accuracy"])
TYPE="fedprox"
fedprox = newDataFrame(seeds, TYPE, degraded=0.2)
TYPE="fedopt"
fedopt = newDataFrame(seeds, TYPE, degraded=0.2)
TYPE="adabest"
adabest = newDataFrame(seeds, TYPE, degraded=0.2)
TYPE="fedeve"
fedeve = newDataFrame(seeds, TYPE, degraded=0.2)
TYPE="afedpd"
afedpd= newDataFrame(seeds, TYPE, degraded=0.2)

# TYPE="fedref"
# root_dir = f"Result/{TYPE}/degrade0.2"

fedref = [fedopt,
          adabest,
          fedeve,
          afedpd,
    # {**pd.read_csv(root_dir+"/fedavg/fedref_cinic10_lda0.1_p4.csv").to_dict("series"), **{"label":"fedref(λ:0.01, ρ:4) Client: avg"}},
    # {**pd.read_csv(root_dir+"/fedprox/fedref_cinic10_lda0.1_p4.csv").to_dict("series"), **{"label":"fedref(λ:0.01, ρ:4) Cient: prox"}},
    ]

newploting(fedavg, fedprox, fedref, title="CINIC-10 Classification", metric=["loss", "accuracy", "f1score"], num=n, thresholds=thresholds, start=0, legend=False)
```
